```python
arr = [1, 2, 2, 3, 4, 4, 5]
freq = {}

for i in arr:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

for i in freq:
    print(i, ":", freq[i])
```

```python
arr = [1, 2, 3, 1]
freq = {}

for i in arr:
    if i in freq:
        print("True")
        break
    else:
        freq[i] = 1
else:
    print("False")
```

```python
arr = [1, 2, 3, 1]
k = 3
freq = {}

for i in range(len(arr)):
    if arr[i] in freq:
        if i - freq[arr[i]] <= k:
            print("True")
            break

    freq[arr[i]] = i
else:
    print("False")
```
