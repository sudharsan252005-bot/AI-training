```python
arr = [1, 4, 3, 2, 6, 5]

n = len(arr)

for i in range(n // 2):
    temp = arr[i]
    arr[i] = arr[n - i - 1]
    arr[n - i - 1] = temp

print(arr)
```

```python
arr = [3, 2, 2, 3]
val = 3

k = 0

for i in range(len(arr)):
    if arr[i] != val:
        arr[k] = arr[i]
        k += 1

print(k)
print(arr[:k])
```
